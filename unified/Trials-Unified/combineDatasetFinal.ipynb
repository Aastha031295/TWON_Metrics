```python
import pandas as pd

# Load datasets
fake_news_df = pd.read_csv("/home/s2shsinh/TWON_Metrics/dataset/fake_news.csv")
hate_speech_df = pd.read_csv("/home/s2shsinh/TWON_Metrics/dataset/hateSpeech.csv")
toxicity_df = pd.read_csv("/home/s2shsinh/TWON_Metrics/dataset/toxicity.csv")

# Keep only the text column and rename label columns to standard names
fake_news_df = fake_news_df[["text", "binary_label"]].rename(columns={"binary_label": "isFake"})
hate_speech_df = hate_speech_df[["text", "label"]].rename(columns={"label": "isHateSpeech"})
toxicity_df = toxicity_df[["text", "toxic"]].rename(columns={"toxic": "isToxic"})

# Find the minimum number of records across all datasets
min_records = min(len(fake_news_df), len(hate_speech_df), len(toxicity_df))

# Take only the minimum number of records from each dataset

```

```python
# Select 2500 records with label 0 and 2500 records with label 1 from each dataset
fake_news_0 = fake_news_df[fake_news_df["isFake"] == 0].head(2500)
fake_news_1 = fake_news_df[fake_news_df["isFake"] == 1].head(2500)
fake_news_df = pd.concat([fake_news_0, fake_news_1])

hate_speech_0 = hate_speech_df[hate_speech_df["isHateSpeech"] == 0].head(2500)
hate_speech_1 = hate_speech_df[hate_speech_df["isHateSpeech"] == 1].head(2500)
hate_speech_df = pd.concat([hate_speech_0, hate_speech_1])

toxicity_0 = toxicity_df[toxicity_df["isToxic"] == 0].head(2500)
toxicity_1 = toxicity_df[toxicity_df["isToxic"] == 1].head(2500)
toxicity_df = pd.concat([toxicity_0, toxicity_1])
```

```python
# Convert isFake values from float to int
fake_news_df["isFake"] = fake_news_df["isFake"].apply(lambda x: int(x) if pd.notnull(x) else x)
```

```python
# Preprocess the text column
fake_news_df["text"] = fake_news_df["text"].apply(preprocess_text)
hate_speech_df["text"] = hate_speech_df["text"].apply(preprocess_text)
toxicity_df["text"] = toxicity_df["text"].apply(preprocess_text)
```

```python
# Add missing columns with default values
fake_news_df["isHateSpeech"] = 2
fake_news_df["isToxic"] = 2

hate_speech_df["isFake"] = 2
hate_speech_df["isToxic"] = 2

toxicity_df["isFake"] = 2
toxicity_df["isHateSpeech"] = 2

# Combine datasets
final_df = pd.concat([fake_news_df, hate_speech_df, toxicity_df], ignore_index=True)

# Save the final dataset to a CSV file
final_df.to_csv("merged_dataset_7_1jan.csv", index=False)

print("Merged dataset created and saved as 'merged_dataset.csv'.")
```
