```python
# Load the dataset
import pandas as pd
file_name = "/home/s2shsinh/TWON_Metrics/dataset/DefaktS_Twitter.binary.csv"  # Replace with your dataset file name
df = pd.read_csv(file_name)

# Separate the dataset into two groups based on the label
label_0 = df[df['binary_label'] == 0]
label_1 = df[df['binary_label'] == 1]

# Randomly sample 2500 records from each group
sample_0 = label_0.sample(n=2500, random_state=42)
sample_1 = label_1.sample(n=2500, random_state=42)

# Combine the samples into a new dataset
balanced_sample = pd.concat([sample_0, sample_1])

# Shuffle the combined dataset
balanced_sample = balanced_sample.sample(frac=1, random_state=42).reset_index(drop=True)

# Save the balanced dataset to a new CSV file
balanced_sample.to_csv("balanced_dataset.csv", index=False)

print("Balanced dataset saved as 'balanced_dataset.csv'.")
```

```python
import pandas as pd

# Load the three datasets
fake_news = pd.read_csv("/home/s2shsinh/TWON_Metrics/dataset/balanced_dataset.csv")  # Replace with the actual file name
hate_speech = pd.read_csv("/home/s2shsinh/TWON_Metrics/dataset/gahd_filtered.csv")
toxicity = pd.read_csv("/home/s2shsinh/TWON_Metrics/dataset/multilingual_toxicity.csv")

# Add a unique label column for each dataset
# Add unique label columns based on their respective labels
fake_news['fake'] = fake_news['binary_label'].apply(lambda x: 1 if x == 1 else 0)
fake_news['hatespeech'] = 0
fake_news['toxicity'] = 0

hate_speech['fake'] = 0
hate_speech['hatespeech'] = hate_speech['label'].apply(lambda x: 1 if x == 1 else 0)
hate_speech['toxicity'] = 0

toxicity['fake'] = 0
toxicity['hatespeech'] = 0
toxicity['toxicity'] = toxicity['toxic'].apply(lambda x: 1 if x == 1 else 0)

# Ensure all datasets have the same columns
fake_news = fake_news[['text', 'fake', 'hatespeech', 'toxicity']]
hate_speech = hate_speech[['text', 'fake', 'hatespeech', 'toxicity']]
toxicity = toxicity[['text', 'fake', 'hatespeech', 'toxicity']]

# Combine the datasets
combined_dataset = pd.concat([fake_news, hate_speech, toxicity], ignore_index=True)

# Save the combined dataset to a new CSV file
combined_dataset.to_csv("combined_dataset1.csv", index=False)

print("Combined dataset saved as 'combined_dataset1.csv'.")
```
